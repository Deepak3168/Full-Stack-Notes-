## 1. What is Software?

**Software** is a set of instructions that tells a computer *what to do*.
Computers themselves are just electronic hardware (chips, wires, memory) —
they don't know how to do anything on their own. Software is the "recipe"
that hardware follows.

There are two broad categories:

| Type | Purpose | Examples |
|------|---------|----------|
| **System software** | Manages the hardware and provides a platform for other software | Operating Systems (Windows, Linux, macOS), device drivers |
| **Application software** | Helps the user do specific tasks | Browsers, games, WhatsApp, Python scripts you write |

```
+-------------------------------------------------+
|                 APPLICATION SOFTWARE             |
|   (Chrome, VS Code, your Python programs, etc.)  |
+-------------------------------------------------+
|                 SYSTEM SOFTWARE                  |
|         (Operating System - Windows/Linux)       |
+-------------------------------------------------+
|                    HARDWARE                      |
|         (CPU, RAM, Disk, Keyboard, Screen)       |
+-------------------------------------------------+
```

**Analogy:** Hardware is like a car's engine and body. Software is the
driver's instructions (accelerate, brake, turn) that make the car actually
go somewhere useful.


## 2. How Computers Execute Programs

A CPU (Central Processing Unit) only understands **machine code** — streams
of 0s and 1s (binary). Every program, no matter what language it's written
in, must eventually become machine code before the CPU can run it.

```
 Your Code (Python/C/Java)
          |
          v
   [ Translation Step ]   <-- compiler / interpreter
          |
          v
     Machine Code (0s and 1s)
          |
          v
   CPU fetches -> decodes -> executes  (repeats millions of times/sec)
```

The CPU follows a simple repeating cycle for every single instruction:

```
   FETCH  --->  DECODE  --->  EXECUTE
     ^                            |
     |____________________________|
        (move to next instruction)
```

- **Fetch**: Get the next instruction from memory (RAM).
- **Decode**: Figure out what operation it represents.
- **Execute**: Actually perform it (add numbers, move data, etc.)

This happens billions of times per second — that's what "GHz" (gigahertz)
in a CPU spec means (billions of cycles per second).


## 3. Compiler vs Interpreter

Both are "translators" that convert human-readable code into something the
CPU can execute — but they work differently.

### Compiler
Translates the **entire program at once** into machine code (or an
intermediate form) *before* it runs. Produces a standalone executable file.

```
Source Code (whole file)
        |
        v
   [ COMPILER ]  --------> Machine Code File (e.g. program.exe)
        |
        v
   Run the .exe directly (fast, no translator needed at run time)
```
Examples: C, C++, Rust, Go

### Interpreter
Translates and executes code **line by line, on the fly**, while the
program is running. No separate executable is produced.

```
Source Code (line 1) -> [ INTERPRETER ] -> Execute line 1
Source Code (line 2) -> [ INTERPRETER ] -> Execute line 2
Source Code (line 3) -> [ INTERPRETER ] -> Execute line 3
                          ... and so on
```
Examples: Python, JavaScript, Ruby

### Quick comparison

| Aspect | Compiler | Interpreter |
|---|---|---|
| Translation | All at once, before running | Line by line, while running |
| Speed of execution | Usually faster | Usually slower |
| Error detection | All errors shown before running | Stops at the first error it hits |
| Output | Standalone executable file | No separate file; needs interpreter every time |
| Examples | C, C++, Rust | Python, JavaScript |

> Python is actually a hybrid — see the next section!


## 4. Python Execution Model

Python is often called "interpreted", but under the hood it does a bit of
both compiling and interpreting:

```
 your_file.py  (source code, human readable)
        |
        v
 [ Python Compiler ]  --- compiles to ---> Bytecode (.pyc, low-level, portable)
        |
        v
 [ Python Virtual Machine (PVM) ]  --- interprets bytecode line by line
        |
        v
     Program Output
```

1. **Compilation step**: Python first compiles your `.py` file into
   **bytecode** — a low-level, platform-independent set of instructions
   (you might have seen `__pycache__` folders with `.pyc` files — that's
   this bytecode, cached so Python doesn't recompile unchanged files).
2. **Interpretation step**: The **Python Virtual Machine (PVM)** then reads
   this bytecode and executes it, instruction by instruction.

This is why Python is called an **interpreted language** in everyday usage,
even though there's a compilation step involved — the key point is that
*you* never see or manage a separate compile step, and there's no final
native `.exe` produced. It "just runs."


````python
# You can literally see Python's compiled bytecode using the `dis` module
import dis

def add(a, b):
    return a + b

dis.dis(add)    

````

## 5. Python vs C

| Feature | Python | C |
|---|---|---|
| Type | Interpreted (compiles to bytecode, then PVM runs it) | Compiled directly to machine code |
| Typing | Dynamically typed (no need to declare variable types) | Statically typed (must declare types) |
| Speed | Slower (extra translation layer at run time) | Faster (runs as native machine code) |
| Memory management | Automatic (garbage collected) | Manual (you allocate/free memory yourself) |
| Syntax | Simple, readable, less code | Verbose, more boilerplate |
| Use cases | Scripting, AI/ML, web backends, automation | OS, embedded systems, drivers, performance-critical apps |

**Same task in both languages — add two numbers:**

Python:
```python
a = 5
b = 10
print(a + b)
```

C:
```c
#include <stdio.h>
int main() {
    int a = 5;
    int b = 10;
    printf("%d", a + b);
    return 0;
}
```

Notice Python needs no type declarations, no semicolons, no `main()`
function, and no compilation step from your side — that's the trade-off:
**less code and faster development**, at the cost of **raw execution
speed**.


````python
# Python example: adding two numbers
a = 5
b = 10
print(a + b)   # Python figures out types automatically
````

## 6. Variables

A **variable** is a named label that refers to a value stored in memory.
Think of it as a sticky note with a name, pointing at a box that holds data.

```
   name           value (stored somewhere in memory)
  +------+        +-----------------+
  | age  |  --->  |       25        |
  +------+        +-----------------+
```

In Python, you don't need to declare a type — just assign a value with `=`.


````python
# Creating variables
name = "Deepak" 
age = 25
height = 5.9
is_student = False

print(name)
print(age)
print(height)
print(is_student)
````

````python
# Variables can be reassigned to a completely different type
x = 10
print(x, type(x))

x = "now I'm a string"
print(x, type(x))
````

## 7. Objects and References

In Python, **everything is an object** — numbers, strings, functions, even
classes. A variable is not a box that "contains" the value directly;
instead, it's a **label/reference that points to an object** somewhere in
memory.

```
  a  ---->  +-------------------+
            |  Object: 10       |
            |  id: 140023...    |
            +-------------------+
                    ^
                    |
  b  -------------->  (b points to the SAME object as a)
```

This matters a lot when multiple variables refer to the *same* object.


````python
a = [1, 2, 3]
b = a          # b now refers to the SAME list object as a, not a copy

b.append(4)

print("a =", a)   # a is affected too, because a and b point to the same object!
print("b =", b)
print(a is b)     # True -> same object in memory
````

````python
# To make an independent copy instead of sharing the reference:
a = [1, 2, 3]
b = a.copy()      # or list(a)

b.append(4)

print("a =", a)   # unaffected
print("b =", b)
print(a is b)      # False -> different objects now
````

**Key takeaway:** `=` in Python doesn't copy data, it copies a
*reference* to the object. Whether that matters depends on whether the
object is **mutable** (can be changed in place, like lists/dicts) or
**immutable** (cannot be changed, like int/str/tuple).

## 9. Type Conversion

Sometimes you need to convert a value from one type to another — this is
called **type casting**. Python gives you built-in functions for this:
`int()`, `float()`, `str()`, `bool()`.

```
   "25"  --int()-->   25        (string to integer)
    25   --str()-->  "25"       (integer to string)
   3.9   --int()-->   3         (float to integer, truncates - doesn't round!)
    0    --bool()--> False      (0 is falsy)
    5    --bool()--> True       (any non-zero number is truthy)
```

This is very common when reading user input (which always comes in as a
string) and you need a number to do math with.


````python
# Implicit vs Explicit conversion

# Implicit (Python does it automatically, no data lost)
x = 5        # int
y = 2.0      # float
z = x + y    # Python auto-converts x to float for the addition
print(z, type(z))

# Explicit (you do it manually)
age_str = "25"
age_int = int(age_str)          # string -> int
print(age_int + 5)

price = 99.99
price_str = str(price)          # float -> string
print("Price: " + price_str)

score = "88"
print(float(score))             # string -> float

print(int(9.99))                # float -> int (TRUNCATES to 9, does not round!)
print(bool(0), bool(1), bool(""), bool("hi"))   # falsy vs truthy values
````

> **Common beginner error:** `int("3.5")` fails! You must go
> `int(float("3.5"))` to convert a decimal-looking string to an int.


````python
try:
    int("3.5")
except ValueError as e:
    print("Error:", e)

print(int(float("3.5")))   # works: "3.5" -> 3.5 -> 3
````

## 10. Operators

Operators are symbols that perform operations on values (called
**operands**).

| Category | Operators | Example |
|---|---|---|
| Arithmetic | `+ - * / // % **` | `7 // 2` → `3` |
| Comparison | `== != > < >= <=` | `5 > 3` → `True` |
| Logical | `and or not` | `True and False` → `False` |
| Assignment | `= += -= *= /=` | `x += 1` |
| Membership | `in`, `not in` | `'a' in "cat"` → `True` |
| Identity | `is`, `is not` | `a is b` |


````python
a, b = 7, 2

print("Addition:      ", a + b)
print("Subtraction:   ", a - b)
print("Multiplication:", a * b)
print("Division:      ", a / b)    # always returns a float
print("Floor Division:", a // b)   # rounds down to nearest whole number
print("Modulus:       ", a % b)    # remainder
print("Exponent:      ", a ** b)   # a to the power of b
````

````python
x = 5
y = 10

print(x == y)   # equal to
print(x != y)   # not equal to
print(x < y)    # less than
print(x >= y)   # greater than or equal to
````

````python
is_sunny = True
is_weekend = False

print(is_sunny and is_weekend)   # both must be True
print(is_sunny or is_weekend)    # at least one must be True
print(not is_sunny)              # flips the value
````

````python
count = 0
count += 1     # same as: count = count + 1
count *= 5     # same as: count = count * 5
print(count)

print('a' in "cat")        # membership
print(5 not in [1, 2, 3])  # membership
````

## 12. `id()` and `type()`

Two very useful built-in functions for inspecting variables:

- **`type(x)`** → tells you the data type of `x`.
- **`id(x)`** → tells you the unique memory address (identity) of the
  object `x` refers to, while the program is running.

```
   x = 10
        |
        v
+------------------+
|  Object: 10       |  <-- id(x) points to this exact memory location
|  type: int        |  <-- type(x) tells you it's an int
+------------------+
```


````python
x = 10
y = "hello"
z = [1, 2, 3]

print(type(x), type(y), type(z))
print(id(x))
print(id(y))
````

````python
# Small integers are CACHED by Python (an internal optimization),
# so two variables with the same small int value can share the same id.
a = 5
b = 5
print(a is b, id(a) == id(b))     # True - both point to the same cached object

# But this is not guaranteed for larger numbers or mutable objects like lists:
c = [1, 2, 3]
d = [1, 2, 3]
print(c is d, id(c) == id(d))     # False - two separate list objects
````

## 13. Basic Memory Concepts

Every value your program creates lives somewhere in **RAM** while the
program runs. Python broadly organizes this into two memory regions:

```
+-----------------------------------------------------+
|                       STACK                          |
|  Stores: function calls, local variable references   |
|  Fast, small, automatically managed (LIFO order)      |
+-----------------------------------------------------+
|                       HEAP                            |
|  Stores: actual objects (lists, strings, custom       |
|  objects, big data)                                   |
|  Managed by Python's Garbage Collector                |
+-----------------------------------------------------+
```

- Variable **names** (references) typically live on the **stack**.
- The actual **objects** they point to live on the **heap**.
- Python uses **automatic garbage collection**: when no variable refers to
  an object anymore, Python automatically frees that memory for you (unlike
  C, where you must manually `free()` memory).

```
 Stack (references)         Heap (actual objects)
 +-------------+            +------------------+
 |  a  ------->|----------->|   [1, 2, 3]       |
 +-------------+            +------------------+
 |  b  ------->|  (b points to the same list as a)
 +-------------+
```


````python
import sys

x = 10
y = "a longer string takes more memory"
z = [1, 2, 3, 4, 5]

print("Size of int 10:      ", sys.getsizeof(x), "bytes")
print("Size of the string:  ", sys.getsizeof(y), "bytes")
print("Size of the list:    ", sys.getsizeof(z), "bytes")
````

## 14. Conditional Statements

Conditionals let your program **make decisions** and run different code
depending on whether something is `True` or `False`.

```
                +------------------+
                |   condition?      |
                +------------------+
                 /              \
             True                False
              |                    |
              v                    v
      run "if" block         check next condition (elif)
                                    |
                              True /  \ False
                                  v      v
                          run "elif"   run "else"
                            block        block
```


````python
age = 20

if age < 13:
    print("Child")
elif age < 20:
    print("Teenager")
else:
    print("Adult")
````

````python
marks = 85

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)

# Nested conditionals
weather = "sunny"
has_umbrella = False

if weather == "rainy":
    if has_umbrella:
        print("Go outside, you're covered!")
    else:
        print("Better stay in, or grab an umbrella.")
else:
    print("Nice day to go outside!")
````

## 15. Loops

Loops let you **repeat** a block of code multiple times, instead of
copy-pasting it.

### `for` loop — when you know how many times / iterating over a sequence

```
 for item in sequence:
      +------------------------+
      |  take next item        |<-------
      |  run loop body          |       |
      +------------------------+       |
              |                        |
       more items left? ----- yes ------
              |
              no
              v
        exit loop
```

### `while` loop — repeats *as long as* a condition stays `True`

```
        +------------------+
   ---->|   condition?      |
   |    +------------------+
   |      True |     | False
   |           v     v
   |     run body   exit loop
   |___________|
```


````python
# for loop - iterating over a range of numbers
for i in range(5):
    print("Iteration:", i)
````

````python
# for loop - iterating over a list
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("I like", fruit)
````

````python
# for loop - iterating over a string
for letter in "Python":
    print(letter)
````

````python
# while loop
count = 0

while count < 5:
    print("Count is", count)
    count += 1     # without this, the loop would run forever!
````

````python
# break and continue
for i in range(10):
    if i == 5:
        break          # stops the loop completely
    print(i)

print("---")

for i in range(10):
    if i % 2 == 0:
        continue       # skips just this iteration, loop keeps going
    print(i)           # only odd numbers get printed
````

````python
# A tiny real-world-style example combining everything so far:
# Simple number guessing check (no randomness, fixed target for demo)

target = 7
guess = int(input("Guess a number between 1 and 10: "))

if guess == target:
    print("Correct!")
elif guess < target:
    print("Too low!")
else:
    print("Too high!")
````
